# 검증 및 측정

* 엄격한 측정과 검증은 성능 최적화뿐만 아니라 신뢰를 구축하고 사용자 기대치에 확실히 부합하는 데 필수적

## 에이전틱 시스템의 측정

* 명확한 목표를 정의하고 유의미한 지표를 수립하며 체계적인 평가 프로세스를 적용함으로써 개발자는 높은 성능과 사용자 만족을 달성하도록 에이전트 시스템의 설계와 구현을 이끌 수 있음

### 측정: 에이전틱 시스템의 핵심

* 효과적인 측정은 에이전트 시스템의 목표와 요구사항에 부합하는 명확하고 실행 가능한 지표를 식별하는 것에서 시작
    * 에이전트의 작업 수행 능력과 사용자 기대 충족 여부를 평가하는 벤치마크 역할
* 성공 여부
    * 사용자 참여 증대
    * 복잡한 프로세스 자동화  
        -> 시스템이 바라는 결과를 반영하는 구체적이고 측정 가능한 목표
* 엄격하고 지속적인 측정이 없다면 다음의 사항들을 알 수 없음
    * 변경 사항이 실제로 개선인지 현실적이고 적대적인 환경에서 에이전트가 어떻게 수행하는지
    * 예상치 못한 회귀를 방지할 수 있는지

* 올바른 지표를 선택하는 일도 중요함
    * 정량적 지표
        * 정확성
        * 응답 시간
        * 견고성
        * 확장성
        * 정밀도
        * 재현율 등
    * 정성적 지표
        * 사용자 만족도 등

-> 시스템이 직면할 현실 세계의 요구를 반영해야 함

* 언어 기반 에이전트의 경우 완전 일치 지표는 실질적인 효용성을 포착하지 못할 때가 많음  
    -> 임베딩 기반 거리, BERTScore, BLEU, ROUGE 같은 의미론적 유사도 측정에 더 의존

* 측정의 이점을 실현하기 위해 평가 메커니즘을 에이전트 개발 생명주기에 직접 통합해야 함
* 평가를 마지막 단계로 미루기 보다 가능한 많은 부분을 자동화해 새로운 코드가 병합되거나 모델이 업데이트될 때마다 테스트 실행  
    -> 다만 자동화된 평가가 모든 것을 말해주지는 않음
* 생소하거나 위험 부담이 큰 도메인에서는 정기적인 샘플링, 사람이 개입하는 검토를 통해 미묘한 문제를 발견, 진행 상황, 남은 과제를 정성적으로 파악할 수 있음

### 개발 라이프사이클에 평가 통합

* 선도적인 팀은  개발의 모든 단계에 자동화된 오프라인 평가를 통합
* 에이전트에 새로운 도구나 워크플로가 추가될 때마다 그에 상응하는 테스트 케이스와 평가 예제를 평가 세트에 추가해야 함  
    -> 시스템의 확장된 기능 범위 전반에 걸쳐 진척도를 측정할 수 있게 함

* 고품질 평가 세트
    * 에이전트가 처리해야 할 작업을 명시하는 살아있는 명세서 역할
    * 재현성과 회귀 감지를 지원  
        -> 과거 결과를 추적 -> 개선이 새로운 오류나 성능 저하를 희생해 얻은 것인지 식별 가능
    * 의사결정에 적량적 토대를 제공

### 평가 세트 생성 및 확장

* 모든 측정 전략의 토대는 고품질 평가 세트
    * 시스템이 현실 세계에서 직면할 다양성, 모호성, 엣지 케이스를 반영해야 함  
        -> 정적이고 수작업으로 선별된 테스트 스위트는 과적합 위험이 있고 롱테일 실패 유형을 놓치며 워크플로와 사용자 행동을 따라잡을 수 없기 때문
* 좋은 평가 세트는 입력 상태와 예상 결과를 모두 정의해 에이전트 행동을 자동으로 검증할 수 있게 함

* 고객 지원 에이전트 예시

```python
{
    "order": {
        "order_id": "A89268",
        "status": "Delivered",
        "total": 39.99,
        "items": [
            {
                "sku": "MUG-001",
                "name": "Ceramic Coffee Mug",
                "qty": 1,
                "unit_price": 19.99
            },
            {
                "sku": "TSHIRT-S",
                "name": "T-Shirt-Small",
                "qty": 1,
                "unit_price": 20.00
            },
        ],
        "delivered_at": "2026-06-11"
    },
    "conversation": [
        {
            "role": "customer",
            "content": "안녕하세요, 배송된 커피 머그잔이 깨져 있습니다. 교환이나 환불이 가능한가요?"
        },
        {
            "role": "assistant",
            "content": "정말 죄송합니다! 파손된 부위의 사진을 찍어 보내주시면 즉시 전액 환불해 드리겠습니다."
        },
        {
            "role": "customer",
            "content": "네, 여기 사진입니다."
        },
    ],
    "expected": {
        "final_state": {
            "tool_calls": [
                {
                    "tool": "issue_refund",
                    "params": {
                        "order_id": "A89268",
                        "amount": 19.99
                    }
                }
            ],
            "customer_msg_contains": ["처리되었습니다", "영업일"]
        }
    }
}
```

* 해당 예제는 여러 가지를 테스트
    * 에이전트가 여러 품목 주문을 올바르게 추론하는지
    * 대화 컨텍스트를 도구 사용과 일치시키는지
    * 사용자에게 친근한 확인 메시지를 생성하는지
* 도구 재현율, 파라미터 정확도, 구문 재현율 같은 평가 지표는 행동을 정량화

* 입력 상태, 대화 이력, 예상 최종 상태를 포함한 구조화된 형식으로 평가 예제를 공식화하면 채점을 자동화하고 다양한 시나리오에 걸쳐 지표를 집계할 수 있음  
    -> 확장성이 좋음

* 팀은 적대적 프롬프팅, 반사실적 편집, 분포 보간과 같은 목표 지향적 생성 기법을 적용할 수 있음  
    -> 미묘한 오류를 발견하고 에이전트 행동의 견고성을 검증

* 실제 데이터에 접근할 수 있는 도메인에서는 도메인 특화 마이닝이 평가 자료의 또 다른 풍부한 원천이 됨
* 도메인 특화 에이전트에 커스텀 벤치마크가 필수적이라 해도 표준 벤치마크는 성능 파악에 도움이 됨

* 잘 구조화된 평가 세트는 단순한 테스트 스위트 이상의 의미를 갖게 되며 에이전트가 처리해야 할 작업에 대한 살아 있는 명세서가 됨  
    -> 평가를 정적인 게이트키핑 기능에서 시스템 개발의 궤적을 직접 형성하는 동적인 모델 주도 피드백 루프로 전환

## 컴포넌트 평가

* 단위 테스트는 소프트웨어 개발의 기본적인 관행, 에이전트 기반 시스템의 개별 컴포넌트를 검증하는 데 매우 중요함
* 잘 설계된 단위 테스트는 시스템의 각 부분이 의도한 대로 작동하는지 확인해 에이전트의 전반적인 신뢰성과 성능에 기여

### 도구 평가

* 도구는 에이전트가 환경에 작용하고 데이터를 조회하거나 변환하며 외부 시스템과 상호작용할 수 있게 하는 핵심 기능
    * 사용 사례를 빠짐 없이 나열
    * 이상적인 시나리오 + 위험한 엣지 케이스 + 숨겨진 가정을 드러낼 수 있는 좋지 못한 시나리오를 포함해야 함

* 성숙한 에이전트 개발 프로세스는 각 도구마다 자동화된 테스트 모음을 정의
* 테스트는 출력의 정확성뿐 아니라 지연시간, 리소스 사용량, 오류 처리까지 명시적으로 검증해 도구가 부하나 장애 상황에서도 우아하게 성능이 저하되도록 보장해야 함

* 도구 테스트는 같은 입력에 대해 출력이 결정적이어야 한다는 점을 검증해야 함
    * 단, 도구 설계 상 확률성이 포함되어 있다면 결정적일 필요 없음 대신 통계적 특성을 점검
* API, DB 등 외부 의존성이 있는 도구는 모킹이나 시뮬레이터를 사용해 엣지 케이스를 재현
* 회귀 테스트는 특히 중요하며 도구가 수정될 때마다 전체 테스트 모음을 다시 실행해야 함

### 계획 능력 평가

* 계획 모듈은 상위 수준의 목표를 실행 가능한 단계 시퀀스로 변환
    * 종종 동적인 의사결정, 분기 로직, 환경 피드백에 대한 적응이 수반
* 에이전트형 계획은 취약하거나 일관성 없는 동작을 피하기 위해 더욱 세심한 테스트가 필요함
* 계획 모듈은 실행 중 학습한 내용에 따라 도구 호출을 순서대로 구성하고 조건문을 조정하며 필요하면 중간에 실행을 중단해야 할 수 있음  
    -> 계획을 검증하는 일은 더 섬세하면서도 더 필수적

* 계획의 품질을 평가하기 위해 먼저 정형화된 워크플로부터 시작
    * 잘 알려진 일반적인 사용자 의도와 검증된 에이전트 응답 쌍으로 구성
* 각 시나리오마다 초기 환경, 대화 이력, 도구 사용과 사용자 커뮤니케이션 측면에서 기대되는 결과를 인코딩

* 플랜을 체계적으로 평가하기 위해 에이전트를 엔드 투 엔드로 실행, 선택된 행동을 추출
* 이후 정답 레이블과 비교 후 여러 자동화된 지표를 계산
    * 도구 재현율
    * 도구 정밀도
    * 파라미터 정확도
* 이러한 지표는 계획 모듈의 작동 과정을 세밀하게 들여다볼 수 있게 해줌

* 평가 지표 함수 정의 예시

```python
def tool_metrics(pred_tools: List[str], expected_calls: List[dict]) -> Dict[str, float]:
    expected_names = [c.get("tool") for c in expected_calls]
    if not expected_names:
        return {"tool_recall": 1.0, "tool_precision": 1.0}
    pred_set = set(pred_tools)
    exp_set = set(expected_names)
    tp = len(exp_set & pred_set)
    recall = tp / len(exp_set)
    precision = tp / len(pred_set) if pred_set else 0.0
    return {"tool_recall": recall, "tool_precision": precision}

def param_accuracy(pred_calls: List[dict], expected_calls: List[dict]) -> float:
    if not expected_calls:
        return 1.0
    matched = 0
    for exp in expected_calls:
        for pred in pred_calls:
            if pred.get("tool") == exp.get("tool") and pred.get("params") == exp.get("params"):
                matched += 1
                break
    return matched / len(expected_calls)
```

* 계획 모듈은 일관성에 대해서도 평가해야 함
* 시간이 지남에 따라 에이전트가 지원해야 하는 전체 범위를 반영하는 계획 시나리오 코퍼스를 축적
* 궁극적으로 계획 능력 평가는 에이전트가 무엇을 해야 하는지 제대로 알고 있는지를 알려줌

### 메모리 평가

* 멀티 턴 대화, 장시간 실행되는 워크플로, 지속적인 사용자 프로필 등에서 연속성과 컨텍스트 인식이 필요한 에이전트에 메모리는 필수 요소
* 메모리 모듈을 테스트하는 일은 단순하지 않음
    * 단순한 저장과 조회의 정확성을 검증하는 것 + 데이터 무결성, 관련성, 효율성까지

* 메모리에 대한 단위 테스트는 먼저 메모리에 기록된 데이터가 정확히 저장되고 시간이 상당히 지난 뒤나 다른 작업이 개입한 뒤에도 정밀하게 다시 조회될 수 있는지 확인해야 함

* 메모리 성능 평가 함수 예시

```python
def evaluate_memory_retrieval(
    retrieve_fn: Any,
    queries: List[str],
    expected_results: List[List[Any]],
    top_k: int = 1
) -> Dict[str, float]:
    """
    k개의 메모리 항목 리스트를 반환하는 검색 함수 `retrieve_fn(query, k)`가 주어졌을 때,
    여러 쿼리에 대해 평가합니다.
    반환값:
      - `retrieval_accuracy@k`: 상위 k개 결과 중 적어도 하나의 예상 항목이 포함된 쿼리의 비율입니다.
    """
    hits = 0
    for query, expect in zip(queries, expected_results):
        results = retrieve_fn(query, top_k)
        # 예상 항목을 검색했는지 확인
        if set(results) & set(expect):
            hits += 1
    accuracy = hits / len(queries) if queries else 1.0
    return {f"retrieval_accuracy@{top_k}": accuracy}
```

* 정확성, 관련성, 효율성, 복원력을 모두 확인해야 함

### 학습 평가

* 학습 컴포넌트는 확률적 특성과 데이터 의존성 때문에 단위 테스트가 가장 복잡함
    * 이미 익힌 행동을 잊어버리는 일이 없도록 보장해야 하기 때문

* 학습을 테스트하는 일은 기본 학습 루프를 검증하는 것에서 시작
* 레이블이 있는 데이터, 피드백, 보상 신호에 따라 에이전트가 자신의 파라미터, 캐시, 규칙을 올바르게 갱신했는가를 확인해야 함

* 지도 학습 에이전트
    * 대표적인 데이터셋으로 학습했을 때 기대한 정확도에 도달하고 검증 데이터에도 제대로 일반화되는지를 단위 테스트로 확인해야 함
* 강화 학습 에이전트
    * 보상 극대화가 시간이 지남에 따라 실제로 행동 개선으로 이어지는지
    * 학습 정체 구간을 감지해 적절히 처리하는지 테스트해야 함

* 일반화는 중요함
    * 테스트는 에이전트가 학습한 행동을 학습 데이터의 분포와 다른 분포를 가지는 시나리오에 얼마나 잘 적용하는지 평가해야 함

* 적응력도 중요함
    * 테스트는 새로운 유형의 사용자 입력, 이전에 보지 못한 도구 실패, 변화하는 보상 구조 같은 분포 변화를 시뮬레이션해 에이전트가 치명적인 망각이나 성능 붕괴 없이 적응할 수 있는지 확인해야 함

* 도구, 계획 능력, 메모리, 학습이라는 이들 컴포넌트를 엄격하게 테스트함으로써 개발자는 에이전트 기반 시스템의 기초 요소가 신뢰성 있게 효과적으로 작동하는 것을 보장할 수 있음

## 총체적 평가

* 통합 테스트는 에이전트 기반 시스템 전체를 평가해 현실적 환경에서 도구, 계획 능력, 메모리, 학습 등 모든 하위 시스템이 매끄럽게 함께 작동하는지 확인
* 통합 테스트는 복잡한 상호작용, 창발적 행동, 엔드 투 엔드 이슈를 드러냄

### 엔드 투 엔드 시나리오에서의 성능

* 통합 테스트의 1차 목표는 실제 사용에 가까운 조건에서 시스템이 시작부터 끝까지 완전한 작업을 수행하는 능력을 검증하는 것
    * 이를 위해 지각, 계획 능력, 도구 호출, 커뮤니케이션 등 에이전트 기반 시스템의 전체 스택을 활용하는 대표 워크플로나 사용자 여정을 구성

* 예시

```python
def evaluate_single_instance(raw: str, graph) -> Optional[Dict[str, float]]:
    if not raw.strip():
        return None
    try:
        ex = json.loads(raw)
        order = ex["order"]
        messages = [to_lc_message(t) for t in ex["conversation"]]
        expected = ex["expected"]["final_state"]

        result = graph.invoke({"order": order, "messages": messages})

        # 어시스턴트의 최종 메시지 출력
        final_reply = ""
        for msg in reversed(result["messages"]):
            if isinstance(msg, AIMessage) and not msg.additional_kwargs.get("tool_calls"):
                final_reply = msg.content or ""
                break

        # 예측된 도구 이름과 인자 수집
        pred_tools, pred_calls = [], []
        for m in result["messages"]:
            if isinstance(m, AIMessage):
                for tc in m.additional_kwargs.get("tool_calls", []):
                    name = tc.get("function", {}).get("name") or tc.get("name")
                    args = json.loads(tc["function"]["arguments"]) if "function" in tc else tc.get("args", {})
                    pred_tools.append(name)
                    pred_calls.append({"tool": name, "params": args})

        # 지표 계산 후 반환
        tm = tool_metrics(pred_tools, expected.get("tool_calls", []))

        return {
            "phrase_recall": phrase_recall(final_reply, expected.get("customer_msg_contains", [])),
            "tool_recall": tm["tool_recall"],
            "tool_precision": tm["tool_precision"],
            "param_accuracy": param_accuracy(pred_calls, expected.get("tool_calls", [])),
            "task_success": task_success(final_reply, pred_tools, expected)
        }
    except Exception as e:
        print(f"[Skipped] 예제가 오류로 실패: {e!r}")
        return None
```

*  중요한 한계는 자동화된 테스트의 푸질이 그에 사용되는 평가 세트와 지표의 품질에 좌우된다는 점
* 평가를 고정된 체크리스트가 아닌 살아 있는 프로세스로 다루는 것이 최선
* 각 평가를 입력 상태부터 에이전트 출력까지의 완전한 상호작용으로 구성하면 시스템이 실제 작업을 얼마나 잘 수행하는지 추적하고 시간에 따른 회귀를 감지하며 계획 능력, 근거 확보, 커뮤니케이션의 취약점을 드러낼 수 있음

### 일관성

* 에이전트 기반 시스템의 일관성은 테스트하기가 특히 어려움  
    -> 확률적이고 비결정적인 파운데이션 모델에 의존하는 경우가 많기 때문
* 일관성 테스트의 초점은 에이전트의 출력이 입력과 정합성을 유지하고 장시간의 교환에서도 일관된 흐름을 이어 가며 사용자가 의도한 질문이나 작업을 안정적으로 처리하는지 확인하는 데 있음

* 일관성 테스트의 핵심 목표 하나느 다양한 시나리오에서 에이전트의 응답이 주어진 입력과 정합성을 유지하는지 검증하는 것

* 긴 상호작용은 추가 복잡성을 야기함  
    -> 시간이 지남에 따라 성능이 저하될 수 있기 때문

* 미묘한 위험은 자동 평가가 드물게 치명적인 엣지 케이스를 놓칠 수 있다는 점

-> 자동화, 수동 리뷰는 이런 과제를 해결하는 데 모두 필수적

* 액터-크리틱 접근법도 일관성 테스트에 유용함

-> 결국 일관성 테스트는 비결정적인 작동을 전제하더라도 에이전트 기반 시스템이 정합적이고 논리적이며 목적 지향적인 출력을 내도록 보장

### 응집성

* 응집성 테스트는 에이전트의 출력이 상호작용 전반에 걸쳐 논리적이고 컨텍스트에 부합하며 일관되게 유지되는지 확인
* 다단계 워크플로를 관리하거나 장기 대화를 지속하는 에이전트에 응집성은 매끄럽고 직관적인 상호작용을 가능하게 함

* 응집성 테스트는 장기 상호작용을 시뮬레이션해 에이전트가 상태에 대한 일관된 이해를 유지하는지 행동이 논리적이며 목표 지향적 순서를 따르는지 검증

-> 응집성 테스트는 에이전트 기반 시스템의 신뢰, 사용성, 실질적 가치를 유지하는데 필수

### 할루시네이션

* 할루시네이션은 에이전트가 틀렸거나 의미가 없거나 지어낸 정보를 생성할 때 발생
    * 정확성과 신뢰성이 최우선이기에 심각한 과제
* 이를 해결하기 위해 에이전트가 현실에 근거한 응답을 일관되게 생성하도록 엄격한 테스트와 완화 전략이 필요함

* 개발자는 신뢰할 수 있는 데이터에 근거해 출력을 고정해야 함
    * 예를 들어 RAG

* 핵심은 콘텐츠 정확성을 보장하는 일
    * 에이전트의 출력이 허구가 아니라 사실 데이터에 기반하는지 검증하는 것을 의미

* 데이터 의존성도 할루시네이션 해결의 핵심 요소
* 피드백 루프는 할루시네이션 탐지와 해결에 필수

* 할루시네이션에 대한 대응은 하이브리드 사람-AI 피드백 루프 중심으로 발전해 옴
* 콘텐츠 정확성에 우선순위를 두고 데이터 의존성을 엄격히 관리하며 피드백 메커니즘을 활용하고 다양한 시나리오에 대해 철저히 테스트하면 할루시네이션 위험을 최소화할 수 있음

### 예기치 않은 입력

* 현실 세계 환경은 예측하기 어려움
* 에이전트는 예상치 못한, 잘못 구성된, 악의적인 입력에도 견고하게 작동해야 함
* 이 영역의 통합 테스트는 학습이나 설계 가정 밖에 있는 입력을 의도적으로 제공

* 효과적인 통합 테스트는 무작위 퍼징만이 아니라 과거 사고나 적대적 분석에서 얻은 통찰로 엣지 케이스를 체계적으로 탐색
