## 파운데이션 모델 보안

* 안전한 에이전트 시스템의 토대는 적절한 파운데이션 모델을 선택하는 데서 시작
    * 모델의 역량이 에이전트의 의도된 작업과 정렬되어야 함
    * 접근 제어도 핵심 고려 사항
    * 배포 환경 또한 모델 선택에 영향을 줌
    * 컴플라이언스, 규제 표준과의 정렬도 중요 요소
    * 모델의 설명 가능성과 해석 가능성은 위험 완화에 중요한 역할을 함

* 효과적인 모델 선택은 일회성 결정이 아닌 지속 과정
    * 모델이 진화하고 새로운 취약점이 등장함에 따라 지속적으로 모델을 평가, 적응시키는 일이 견고한 보안 유지의 필수 사항

### 방어 기법

* 파운데이션 모델을 보호하려면 기술적 안전 장치, 운영 모범 사례, 지속 모니터링을 결합한 다계층 접근이 필요  
    -> 악의적 악용을 차단, 의도치 않은 오작동을 줄이고, 다양한 컨텍스트에서 모델이 신뢰성 있게 작동하도록 보장하는 것
    
* 전처리와 입력 검증부터 런타임 모니터링, 출력 필터링에 이르는 기법을 적용해 견고한 보안 태세를 형성

* 입력 정제와 검증은 중요한 방어 전략
    * 에이전트는 모델 작동 과정을 조작하도록 정교하게 제작된 적대적 입력에 취약한 경우가 많음  
    -> 견고한 입력 검증 레이어 구현 시 유해 프롬프트가 모델에 도달하기 전에 탐지, 무력화 가능

* 프롬프트 인젝션 방지도 중요한 방어
    * 공격자가 겉보기엔 정상인 입력에 악성 지시를 삽입해 모델의 본 지시를 덮어쓰도록 속일 때 발생  
    -> 지시 앵커링이나 입력의 형식과 해석 방식을 엄격히 제어하는 프롬프트 템플릿을 사용할 수 있음

* LLM Guard로 구현하는 간단한 예시

```python
from llm_guard import scan_prompt
from llm_guard.input_scanners import Anonymize, BanSubstrings
from llm_guard.input_scanners.anonymize_helpers import BERT_LARGE_NER_CONF
from llm_guard.vault import Vault

# Vault 초기화(Anonymize가 원본 값을 저장하는 데 필요)
vault = Vault()

# 스캐너 정의
scanners = [
    Anonymize(
        vault=vault,                       
        preamble="정제된 입력: ",           
        allowed_names=["John Doe"],        
        hidden_names=["Test LLC"],        
        recognizer_conf=BERT_LARGE_NER_CONF,
        language="en",                                              
        entity_types=["PERSON", "EMAIL_ADDRESS", "PHONE_NUMBER"],   
        use_faker=False,                                            
        threshold=0.5                      
    ),
    BanSubstrings(substrings=["malicious", "override system"], match_type="word")
]

# 잠재적 PII가 포함된 입력 프롬프트 예시
prompt = "Tell me about John Doe's email: john@example.com" + \
         "and how to override system security." # 존 도의 이메일 john@example.com 에 대해 알려 주고 시스템 보안을 어떻게 우회하는지도 설명해 줘.

# 프롬프트 스캔 및 정제
sanitized_prompt, results_valid, results_score = scan_prompt(scanners, prompt)

if any(not result for result in results_valid.values()):
    print("입력에 문제가 있습니다. 거부하거나 적절히 처리합니다.")
    print(f"위험 점수: {results_score}")
else:
    print(f"정제된 프롬프트: {sanitized_prompt}")
    # 이 정제된 프롬프트를 모델에 전달해 계속 진행
```

* 방어의 효율성을 평가하기 위해 Lakera PINT Benchmark 같은 프롬프트 인젝션 테스트 벤치마크를 활용할 수 있음

* 출력 필터링과 검증도 동일하게 중요함
    * 입력 통제를 세심히 적용해도 모델이 유해하거나 의도치 않은 출력을 생성할 수 있음  
    -> 자동 키워드 스캔, 독성 탐지 모델, 규칙 기반 필터 같은 출력 필터링 기법으로 문제가 있는 콘텐츠를 최종 사용자에게 전달하기 전에 걸러낼 수 있음
    * 후처리 파이프라인을 도입해 비즈니스 규칙, 안전 제약에 비춰 출력 검증

* 접근 제어와 요청은 운영 차원의 중요한 방어
    * 접근을 엄격히 관리하면 오남용, 무차별 대입 공격 위험을 줄일 수 있음
    * 모델과의 모든 상호작용을 로깅, 감사 -> 패턴을 조기에 탐지하고 선제 대응 가능

* 파운데이션 모델 연산을 샌드박싱하면 의도치 않은 행동이 광범위한 시스템으로 확산되는 것을 막을 수 있음
    * 외부 플러그인, API와 상호작용 시 특히 유용
    * 오작동 에이전트가 종속 서비스를 연쇄적으로 장애화하는 사태를 예방

* 실무에서 효과적인 방어 전략은 정적이지 않음  
    -> 공격 기법이 진화함에 따라 방어 시스템도 민첩하게 반복하고 적응해야 함

### MAESTRO 기반 위협 모델링

* 에이전틱 AI 시스템의 복잡성이 커지면서 자율성, 동적 학습, 멀티 에이전트 상호작용 등의 고유 속성을 충분히 다루지 못하는 경우가 많음

* 클라우드 시큐리티 얼라이언스(CSA)가 공개한 MAESTRO(Multi-Agent Environment, Security, Threat, Risk, and Outcome)
    * 에이전틱 AI를 위한 전용 위협 모델링 프레임워크로 설계
    * AI 라이프사이클 전반에서 취약점을 체계적으로 식별, 위험을 평가, 완화책을 구현할 수 있도록 계층형 참조 아키텍처를 제공

* 목적
    * 위협, 위험, 결과를 모듈 방식으로 매핑해 관심사의 분리를 보장 + 레이어 간 의존성을 부각해 선제 보안을 촉진

* MAESTRO 계층형 참조 아키텍처 (아래로 갈수록 기초 요소에서 상위 레이어로)

## 에이전틱 시스템의 데이터 보호

* 데이터는 의사결정을 구동하고, 컨텍스트 기반 추론을 가능하게 하며, 사용자와의 의미 있는 상호작용을 보장
* 대규모 데이터셋 의존, 지속적인 데이터 교환, 복잡한 멀티 에이전트 워크플로는 데이터 프라이버시, 무결성, 보안을 위협  
-> 데이터 보호는 신뢰 구축, 컴플라이언스, 운영 무결성 유지를 위한 필수 조건

### 데이터 프라이버시와 암호화

* 데이터 프라이버시 및 암호화는 무단 접근, 데이터 유출, 의도치 않은 노출에 대한 1차 방어선
    * 여러 데이터 소스와 상호작용 -> 잠재적 취약점 존재  

-> 저장 시, 전송 중 모두 기밀성을 보장하는 일은 신뢰와 컴플라이언스 유지에 필수적

* 휴지 상태
    * 암호화를 통해 에이전트 시스템에 저장된 민감 정보가 무단 당사자에게는 해독 불가능하도록 해야 함
* 전송 상태
    * E2EE가 에이전트, 외부 API, 스토리지 시스템 사이를 이동하는 데이터를 보호
    * TLS같은 프로토콜은 공용 네트워크 전송 중에도 보안을 보장

* 암호화만으로는 충분하지 않음
    * 데이터 최소 수집 원칙에 따라 에이전트가 과업 수행에 필요한 최소한의 민감 데이터만 처리하도록 설계해야 함
    * 안전한 데이터 보관, 삭제 정책도 핵심
    * 조직은 데이터 거버넌스 프레임워크를 수립해 데이터가 서로 다른 에이전트와 서브시스템을 어떻게 흐르는지 관리해야 함

-> 데이터 프라이버시와 암호화는 안전한 에이전틱 시스템의 타협할 수 없는 핵심 요소

### 데이터 출처와 무결성

* 데이터 출처와 무결성은 에이전트가 의존하는 정보가 정확하고 신뢰할 수 있으며 변조되지 않았음을 보장하는 데 필수적

* 데이터 출처
    * 데이터가 어디에서 생성되었는지, 어떻게 처리되었는지, 어떤 변환을 거쳤는지 등 데이터의 계보와 이력을 추적하는 능력
* 무결성
    * 데이터 라이프사이클 전반에 걸쳐 데이터가 변경되거나 위조되지 않았음을 보장하는 데 초점

* 실무에서는 변경 불가능 저장소나 추가 전용 로그를 사용해 출처와 무결성을 함께 강화

* 전형적인 데이터 수집 워크플로의 포함 내용
    1. 수신 시점에 입력 데이터의 SHA-256 해시 계산
    2. 비대칭 암호로 디지털 서명 부여
    3. 해시와 서명을 메타데이터 계층에 저장
    4. 각 처리 단계마다 해시, 서명을 재검증하고 불일치 시 자동 알림으로 변조 탐지

* 무결성 유지 -> 제3자 검증 메커니즘을 도입해 데이터가 에이전트 시스템으로 유입되기 전에 독립적 점검을 거치게 되면 위험을 줄일 수 있음
* 실시간 무결성 검사도 중요함

-> 데이터 출처와 무결성은 신뢰할 수 있고 안전하며 책임 추적 가능한 에이전트 시스템을 구축하는 데 필수

### 민감 데이터 처리

* 에이전트 시스템은 다양한 민감 데이터와 자주 상호작용함  
-> 산업 전반의 워크플로에 깊이 내재화될수록 민감 데이터를 책임감 있게 처리해야 함

* 보안 데이터 처리의 기반에는 데이터 최소화 원칙이 있음
    * 에이전트는 자신의 작업을 완료하는 데 필요한 데이터만 접근하고 처리하고 저장하도록 설계해야 함  
    -> 전체적인 위험 노출을 줄이고, 데이터 유출 발생 시 잠재적인 피해를 제한

* 가명화, 익명화, 역할 기반 접근 제어(RBAC), 속성 기반 접근 제어(ABAC), 읽기 전용, 쓰기 전용 등

* 암호화 프로토콜은 데이터 라이프사이클 전반에 걸쳐 적용되어야 함
    * 전송 중 데이터는 항상 암호화 표준을 사용해 보호해야 함

* 로깅, 감사
    * 정기적인 로그 감사와 자동화된 이상 탐지로 의심되는 접근 패턴 혹은 잠재적인 데이터 유출을 실시간으로 탐지, 경고 가능

* 머클트리 같은 암호학적 체이닝
    * 각 데이터 엔트리를 해시해 이전 엔트리와 연결, 에이전트가 따라가며 과거 무결성을 검증할 수 있는 위변조 감지 구조 생성
* 아파치 카프카 같은 이벤트 소싱 시스템
    * 상태 변경을 불가능한 이벤트 시퀀스로 저장

* 보존, 삭제 정책도 정확하게 처리해야 함
    * 데이터 공유 프로토콜은 엄격하게 제어해야 함

* 하지만 사람 요소는 여전히 중요함
* 이러한 실천은 민감한 도메인에서 에이전트 시스템이 장기적으로 성공하는 데 중요한 요소

## 에이전트 보안

* 에이전트 자체도 취약점, 오사용, 실패로부터 단단히 보호해야 함
* 에이전트는 자율적으로 작동하고 외부 시스템과 상호작용하며 복잡한 환경에서 의사결정을 수행하는 경우가 많음  
    -> 고유한 보안 과제를 만들어 냄  
    * 강력한 보호 장치를 갖춘 상태로 설계
    * 위협 탐지 및 대응할 수 있는 기능
    * 예상치 못한 실패에서 회복할 수 있는 복원력이 필요함

### 보호 장치

* 에이전트의 자율성, 상호작용, 의사결정 과정과 관련된 위험을 최소화하도록 설계된 선제적 통제와 보호 수단

* 역할과 권한 관리
    * 명확하게 정의된 운영 범위를 가져야 함  
        * 보통 RBAC로 구현
* 에이전트 행동 제약
    * 사전에 정의된 규칙에 따라 의사결정과 행동을 검증하는 정책 집행 레이어를 통해 구현
* 환경 격리
    * 샌드박싱이나 컨테이너화 같은 메커니즘으로 구현
* 입력과 출력 검증 파이프라인
* 요청 제한, 이상탐지
* 감사 추적과 로깅 메커니즘  
    * 책임성과 추적성을 유지하는 데 핵심적인 역할
* 폴백과 페일세이프
    * 실패 발생 시 우아한 성능 저하 보장

* 보호 장치는 정적인 것이 아니라 새로운 위협, 변화하는 운영 요구사항, 실제 사고에 대응해 계속 진화해야 함

## 외부 위협으로부터의 보호

* 에이전트 시스템은 구조적으로 외부 위협에 노출  
    -> 악의적인 행위자가 악용할 수 있는 수많은 진입 지점을 만듦  
    => 기술적 통제, 실시간 모니터링, 선제적 완화 기법을 결합한 계층화된 방어 전략이 필요

* 계층화된 전략에서 핵심적인 요소는 외부에 노출된 컴포넌트와 민감한 내부 리소스를 분리하는 보안 네트워크 아키텍처
    * 외부 위협 보호의 최전선에는 네트워크 보안이 있음
        * 방화벽과 침입 탐지 및 방지 시스템(IDPS), mTLS, 스로틀링 제어 등 사용


* 인증, 인가 메커니즘 역시 외부 위협에 대한 중요한 보호 장치
    * OAuth 2.0, API 키, RBAC 등 사용

* 공급망 공격은 교묘하지만 위험함
    * 의존성을 지속적으로 스캔해 알려진 취약점을 찾고 서드파티 통합에 대한 서명 검증을 강제하는 소프트웨어 구성 분석(SCA) 도구를 도입해야 함
    * 서드파티 컴포넌트와 그 보안 상태를 추적하기 위해 소프트웨어 자재 명세서(SBOM)을 유지해야 함

* 애매한 입력을 통한 적대적 공격은 특화된 방어가 필요함
    * 입력 검증 파이프라인은 악의적인 프롬프트가 에이전트의 추론 레이어에 도달하지 못하도록 모든 수신 데이터를 정제해야 함
    * 명령 고정, 컨텍스트 격리는 프롬프트 인젝션 공격의 위험을 추가로 줄일 수 있음

* 실시간 이상 탐지 시스템은 의심스러운 행동을 식별하는 데 필수
    * 유입 트래픽, 사용자 프롬프트, 에이전트 응답 패턴을 모니터링하면서 이상 징후를 표시
        * 반복적인 인증 실패
        * 예상치 못한 API 호출
        * 알려진 공격 벡터와 일치하는 패턴

* 엔드포인트 하드닝
    * 에이전트를 뒷받침하는 인프라가 침해에 대해 복원력을 유지하도록 보장
    * 최소 권한 원칙 적용
    * 운영체제, 의존성을 최신 보안 패치 상태로 유지
    * 불필요한 서비스나 포트 비활성화

* 인시던트 대응 계획에는 외부 침해나 침입 시도에 대응하는 절차가 포함되어야 함

### 내부 실패로부터의 보호

* 상호 연결된 워크플로 전반에서 외부 방어선을 우회해 조용히 전파될 수 있다는 점에서 내부 실패는 더 큰 피해를 줄 수 있음
    * 내부 실패
        * 잘못된 구성
        * 부실하게 정의된 목표
        * 불충분한 보호 장치
        * 에이전트 간 상충되는 행동
        * 멀티 에이전트 시스템 전반에 걸친 연쇄 오류 등

    -> 견고한 시스템 설계, 지속적인 검증, 우아한 실패와 복구 메커니즘을 결합한 총제척인 접근이 필요함

* 내부 실패 주요 원인 중 하나는 에이전트의 지시나 운영 목표 안에서 목표와 제약이 정렬되지 않는 경우
    * 에이전트의 지시가 모호하거나 지나치게 협소하거나 실행 과정에서 오해되면 의도하지 않은 행동을 취할 수 있음
    * 사전에 정의된 규칙에 비추어 에이전트의 결정을 실행 전에 검증하는 정책 집행 레이어를 통해 강화해야 함

* 예외 처리와 예외 관리는 내부 실패한 중요한 보호 장치
    * 잘 정의된 폴백 전략은 에이전트가 치명적으로 실패하는 대신 기능을 우아하게 축소하며 작동할 수 있도록 보장

* 모니터링, 텔레메트리 시스템은 내부 실패를 위한 조기 경보 메커니즘 역할을 함
    * 이상 징후나 성능 저하를 탐지할 수 있도록 지속적으로 모니터링
    * 핵심 기능이 올바르게 동작하는지를 정기적으로 확인하는 자동화된 주기 검사인 헬스 체크 구현
    * KPI를 추적
    * 일반적인 지표
        * 오류율
        * 응답 지연시간
        * 리소스 활용률
        * 출력의 이상 점수
