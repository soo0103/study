# 에이전트 시스템을 위한 UX 디자인

* UX는 성공을 좌우하는 핵심 요소
* 에이전트의 효과성, 신뢰성, 도입을 결정하는 건 사용자가 에이전트와 상호작용하는 방식
* 에이전트는 텍스트, 그래픽, 인터페이스, 음성, 비디오 등 다양한 모달리티로 상호작용 가능

* 에이전트 모달리티 유형 비교

    | 모달리티 | 보편성 | 예시 활용 사례 | 적합한 상황 |
    |-----|-----|-----|-----|
    | 텍스트 | 매우 흔함 | 고객 지원 챗봇, 생산성 어시스턴트 | 명확하고 비동기적이며 검색 가능한 소통이 필요할 때 |
    | GUI | 흔함 | 워크플로 오케스트레이션 대시보드, AI 코딩 어시스턴트 | 시각적 구조, 컨텍스트 관리, 다단계 워크플로가 중요한 경우 |
    | 음성 | 덜 흔함 | Siri, Alexa, Google Home, 콜센터 자동화 | 손을 쓰지 않는 상호작용이나 자연스러운 대화가 필요한 경우 |
    | 비디오 | 드묾 | 가상 튜터, 의료 상담 아바타, 대화형 학습 에이전트 | 시각적 시연, 풍부한 표현, 몰입형 학습이 필요한 경우 |

* 또 다른 핵심 UX 고려 사항은 시간이 길어질 경우 컨텍스트를 어떻게 관리하는지임
* 이러한 능력이 없으면 기술적으로 뛰어나도 단절되고 반응이 둔한 느낌을 줌
* 사용자는 에이전트가 무엇을 할 수 있고 할 수 없는지 언제 개입하거나 지침을 제공해야 하는지 알아야 함

* 신뢰와 투명성은 에이전트 시스템의 긍정적 사용자 경험을 떠받치는 토대
* 예측 가능한 행동과 명확한 행위 설명은 사용자가 고위험 시나리오에서도 에이전트를 신뢰하고 의존하도록 도움

## 모달리티

* 에이전트 시스템은 다양한 모달리티로 사용자와 상호작용
* 모달리티마다 고유한 강점, 한계, 설계 고려 사항이 있음

### 텍스트 기반 인터페이스

* 에이전트 시스템과 상호작용하는 가장 흔하고 범용적인 방식
* 단순성, 친숙함, 기존 워크플로에 쉽게 통합되는 점이 널리 채택되는 이유
* 동기식 대화, 비동기식 상호작용을 모두 지원
* 문제 발생 시 투명성, 책임성, 트러블 슈팅이 쉬워짐

* 텍스트 기반 인터페이스의 핵심 한계는 발견 가능성
* 사용자가 에이전트가 지원하는 능력이나 효과적인 명령 표현 방식을 모르는 경우가 많음
* 행동 유도성이 부족하면 혼란이 생기고 에이전트 기능이 충분히 활용되지 않으며 요청이 지원 범위를 벗어날 때 사용자의 불편함이 커짐

* 효과적인 텍스트 기반 에이전트를 설계하려면 발견 가능성을 높이는 전략이 필요함
* 사용자가 에이전트의 작동 범위를 이해하도록 해 시행착오를 줄임

* 추가로 명확성, 컨텍스트 유지, 오류 관리에 대한 세심한 주의가 필요함

### 그래픽 인터페이스

* 그래픽 인터페이스는 텍스트, 버튼, 아이콘 등 여러 그래픽 요소를 결합해 사용자와 에이전트 시스템이 시각적인 상호작용을 하도록 함

* 정보를 시각적으로 제시해 인지 부하를 줄인다는 강점이 있음
* 잘 설계한 인터페이스는 직관적이고 이해하기 쉬운 형식으로 보여줌

* 새롭게 떠오르는 분야는 생성형 UI의 등장
* 사용자 질의에 따라 인터페이스 요소, 데이터 시각화, 구조화된 출력을 동적으로 만듦

* 화면의 물리적 공간이 한정적
* 중요한 정보를 노이즈 속에 파묻지 않도록 우선순위를 정해야 함
* 에이전트는 인터페이스 반응성을 관리해야 함

* 자동화와 사용자 통제의 균형도 중요함
* 그래픽 인터페이스는 에이전트의 자율성과 사용자 주도 행동을 자주 혼합함

### 음성 인터페이스

* 에이전트 시스템과 자연스럽고 손을 쓰지 않고 상호작용하도록 제공
* 시각 장애, 이동성이 제한된 사용자에게 접근 가능한 대안을 제공해 에이전트 시스템의 포용성을 높임

* 실시간으로 음성을 처리하려면 음성 인식, 의도 인식, 적절한 응답 생성이 필요함

* 중단을 우아하게 처리하는 능력은 설계에서 특히 중요함
* 효과적인 음성 에이전트는 유연성을 반영해야 함

* 현대의 음성 에이전트는 명령을 파싱하고 정적인 답을 반환하는 수준을 넘어 외부 컨텍스트 찾모, 기록 갱신, 실시간 액션 수행을 할 수 있음

* 오픈 AI의 실시간 음성 API를 사용하는 FastAPI 서버 예시

```python
import os, json, base64, asyncio, websockets
from fastapi import FastAPI, WebSocket
from dotenv import load_dotenv

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

VOICE = "Coral" # 목소리 선택
PCM_SR = 24000 # 클라이언트 사이드에서 사용할 새플레이트(16kHz)
PORT = 5050

app = FastAPI()

@app.websocket("/voice")
async def voice_bridge(ws: WebSocket) -> None:
    """
    1. 브라우저가 ws://host:5050/voice로 WebSocket 연결
    2. OpenAI Realtime API (wss://api.openai.com/v1/realtime)에 연결 및 세션 초기화
    3. from_client(): 브라우저 -> OpenAI로 PCM 오디오 스트리밍(input_audio_buffer.append)
    4. to_client(): OpenAI -> 브라우저로 어이스턴트 음성 델타 전달 (response.audio.delta)
    5. 인터럽션 처리: 사용자 발화 감지 시 어시스턴트 응답 즉시 중단 (conversation.item.truncate)
    """
    await ws.accept()

    # OpenAI Realtime API에 하위 연결 생성
    openai_ws = await websockets.connect(
        "wss://api.openai.com/v1/realtime?model=gpt-realtime-mini",
        additional_headers={
            "Authorization": f"Bearer {OPENAI_API_KEY}",
            "OpenAI-Beta": "realtime=v1"
        },
        max_size=None,
        max_queue=None # 데모 단순화를 위해 무제한 설정
    )

    # Realtime 세션 초기화
    await openai_ws.send(json.dumps({
        "type": "session.update",
        "session": {
            "turn_detection": {"type": "server_vad"},
            "input_audio_format": f"pcm16",
            "output_audio_format": f"pcm16",
            "voice": VOICE,
            "modalities": ["audio"],
            "instructions": "당신은 간결한 AI 에이전트입니다. 한국어로 예의 있는 말투로 대답하세요. 최대한 짧게 대답하세요."
        }
    }))

    last_assistant_item = None # 현재 어시스턴트 응답 추적
    latest_pcm_ts = 0 # 클라이언트로부터의 ms 타임스탬프
    pending_marks = []

    async def from_client() -> None:
        """브라우저에서 OpenAI로 마이크 PCM 청크를 중계"""
        nonlocal latest_pcm_ts
        async for msg in ws.iter_text():
            data = json.loads(msg)
            pcm = base64.b64decode(data["audio"])
            latest_pcm_ts += int(len(pcm) / (PCM_SR * 2) * 1000) # 단순 ms 카운터
            await openai_ws.send(json.dumps({
                "type": "input_audio_buffer.append",
                "audio": base64.b64encode(pcm).decode("ascii")
            }))

    async def to_client() -> None:
        """어시스턴트 오디오 중계 및 인터럽션 처리"""
        nonlocal last_assistant_item, pending_marks
        async for raw in openai_ws:
            msg = json.loads(raw)

            # 어시스턴트가 말하는 경우
            if msg["type"] == "response.audio.delta":
                pcm = base64.b64decode(msg["delta"])
                await ws.send_json({"audio": base64.b64encode(pcm).decode("ascii")})
                last_assistant_item = msg.get("item_id")
            
            # 사용자가 말하기 시작 -> 어시스턴트 발화 중단
            if msg["type"] == "input_audio_buffer.speech_started" and last_assistant_item:
                await openai_ws.send(json.dumps({
                    "type": "conversation.item.truncate",
                    "item_id": last_assistant_item,
                    "content_index": 0,
                    "audio_end_ms": 0 # 즉시 중단
                }))
                last_assistant_item = None
                pending_marks.clear()

    try:
        await asyncio.gather(from_client(), to_client())
    finally:
        await openai_ws.close()
        await ws.close()

if __name__ == "__main__":
    import uvicorn
    uvicorn.run("realtime_voice_minimal:app", host="0.0.0.0", port=PORT)
```

* 음성 인터페이스는 별도의 조작이 필요 없는 작업, 빠른 질의, 액션 중심 워크플로에서 가장 효과적

### 비디오 기반 인터페이스

* 시각, 청각, 텍스트 요소를 하나의 일관된 경험으로 결합하는 새로운 에이전트 상호작용 양식
* 비디오 인터페이스의 핵심 강점은 시각적 단서, 음성, 텍스트 오버레이, 애니메이션 등 여러 감각 채널을 결합해 더 풍부하고 표현력 있는 상호작용을 제공한다는 점
* 고품질 비디오 상호작용에는 상당한 연산 성능과 대역폭이 필요함

### 모달리티 결합을 통한 매끄러운 경험

* 가장 매력적인 에이전트 UX는 여러 모달리티를 하나의 응집된 사용자 여정으로 결합함
* 모달리티를 넘나들면서 상태와 컨텍스트를 유지하는 능력은 훌륭한 에이전트 시스템 설계의 핵심
* 모달리티 유동성을 설계하려면 정보, 작업 진행 상황, 사용자 선호가 전환 과정에서 유실되지 않도록 세심한 상태 관리와 컨텍스트 지속성이 필요함

### 자율성 슬라이더

* 매우 중요한 차원은 에이전트에 부여하는 자율성의 수준
* 효과적인 에이전트 시스템은 완전 수동 제어에서 부분 자동화, 완전 자율 운영까지 에이전트의 자율성을 사용자가 부드럽게 조정할 수 있어야 함
* 자율성 슬라이더라 부르는 이 개념은 사용자가 특정 시점에 유지할 통제와 위임할 범위를 선택할 수 있게 함

* 사용자, 작업, 컨텍스트에 따라 요구되는 에이전트 자율성의 정도는 다름
* 이러한 선호가 고정되지 않는다는 것이 중요한 점
* 사용자 신뢰, 작업 친숙도, 위험도, 업무량에 따라 변화함

* 수동
    * 개발자가 에이전트 도움 없이 모든 코드를 직접 작성
* 요청
    * 에이전트가 코드 자동완성, 리팩터링, 문서 스니펫을 선제적으로 제안하지만 적용 전에는 개발자가 각 제안을 검토해 수락
* 완전 자율
    * 에이전트가 표준 리팩터링 적용, 린터 오류 수정, 프로젝트 규약에 따른 보일러플레이트 파일 생성 등 특정 작업을 개별 승인 없이 자율적으로 수행

* 자율성 슬라이더를 효과적으로 통합하려면 다음 설계 원칙을 고려해야함
    * 자율성의 단계 노출을 명확히 하라
    * 매끄러운 전환을 가능하게 하라
    * 각 수준에서 예측 가능하고 투명한 기능을 제공하라
    * 각 수준의 위험과 이점을 소통하라
    * 사용자 신뢰와 역량에 따라 자율성을 조절하라

## 동기식 대 비동기식 에이전트 경험

* 동기식 에이전트는 사용자와 실시간으로 상호작용하며 즉각적으로 대화를 진행함
* 비동기식 에이전트는 사용자와 독립적으로 작동하고 시간 간격을 두고 이따금씩 소통

* 동기식과 비동기식 디자인의 선택은 작업의 성격, 사용자 기대, 운영 컨텍스트에 크게 좌우됨
    * 동기식 에이전트는 즉각적 피드백이나 실시간 의사결정이 필요한 작업에 강함
    * 비동기식 에이전트는 오래 걸리거나 백그라운드 처리가 필요하거나 사용자의 지속적 주의를 요구하지 않는 워크플로에 더 적합

### 동기식 경험을 위한 디자인 원칙

* 즉시성, 명확성, 반응성이 중요함
* 동기식 환경의 에이전트는 응답에서 명료함과 간결함을 우선해야 함
* 동기식 디자인에서도 오류 처리가 중요함

### 비동기식 경험을 위한 디자인 원칙

* 즉시성보다 유연성, 지속성, 장기적 명료성을 우선함
* 작업 상태와 결과를 명확히 전달해야 함
* 컨텍스트 관리도 중요한 디자인 원칙
* 사용자 기대를 효과적으로 관리해야 함

### 능동적 개입과 침해적 행동 사이의 균형

* 에이전트가 언제, 어떻게 능동적으로 사용자를 참여시켜야 하는지 결정하는 건 항상 어려운 일
* 능동성의 균형은 컨텍스트 인식과 사용자 통제에 달림
* 에이전트가 능동적으로 연락할 때는 관련성을 최우선해야 함

## 컨텍스트 유지와 연속성

* 컨텍스트 유지와 연속성을 보장하는 것은 효과적인 에이전트 시스템을 설계하는 데 중요함
* 사용성, 효율, 신뢰에 직접적인 영향을 줌

* 컨텍스트 유지는 기술적 능력이지만 UX 관점의 문제
* 구현 방식은 UX를 직접 규정
* 효과적인 컨텍스트 유지를 위해서는 단기오아 장기 메모리를 적절히 관리해야 함

* 컨텍스트 관리는 데이터 지속성, 프라이버시 문제, 메모리 한계를 신중히 다뤄야 함
* 과도하게 보존하거나 불필요한 세부를 저장하면 비효율적이 되거나 프라이버시를 침해할 위험이 있음

### 상호작용 간 상태 유지

* 상태 관리는 컨텍스트 연속성을 떠받치는 중요한 역할을 함
* 사용자가 무엇을 달성하려 하는지, 다음 논리적 단계가 무엇인지 정확히 추적해야 함

* 효과적인 상태 관리는 시스템이 사용자나 세션을 어떻게 식별, 추적하느냐에 달려있음
    * 로그인 사용자의 경우 상태를 사용자 계정에 직접 연결해 기기와 세션을 넘나드는 메모리 지속성을 제공
    * 익명 상호작용에서는 보통 쿠키나 토큰 등의 세션 식별자가 필요함

* 에이전트 시스템이 확장되면 상태를 db나 분산 캐시에 유지해 연속성을 보장하고 로드 밸런싱을 지원하며 멀티디바이스 경험을 가능하게 함

* 효과적인 상태 유지는 명확한 세션 경계, 데이터 검증, 폴백 메커니즘을 요구
* 상태 관리를 잘 구현하면 에이전트는 불필요한 반복 없이 복잡한 작업을 안내하고 인지 부하를 줄이며 협업이 지속된다는 느낌을 줌

### 개인화와 적응성

* 개인화는 단순한 컨텍스트 기억을 넘어 과거 상호작용과 선호를 사용해 에이전트의 행동, 응답, 추천을 사용자별로 조정
* 개인화는 여러 형태로 나타남
    * 선호 보존
        * 알림 선호, 자주 선택하는 옵션 같은 사용자 설정을 기억
    * 행동 적응
        * 관찰된 사용자 패턴에 기반해 응답 스타일이나 상호작용 흐름을 조정함
    * 능동적 지원
        * 과거 행동을 바탕으로 사용자 필요를 예측하고 제안을 제공함

* 개인화는 어떤 데이터를 저장하고 어떻게 사용하는지에 대한 투명한 소통 등 프라이버시 문제를 신중히 다뤄야 함

## 에이전트 능력 커뮤니케이션

* 효과적인 에이전트 경험을 설계할 때 중요한 과제 하나는 사용자가 에이전트가 무엇을 할 수 있으며 어떻게 상호작용해야 하는지 명확히 이해하게 하는 일
* 백엔드 에이전트 설계가 지원 기능을 결정하지만 발견 가능성, 직관성, 실사용성을 갖추는지는 사용자 경험이 좌우함

* 효과적인 에이전트 UX는 인터페이스 자체를 통해 능력을 선제적으로 커뮤니케이션해 이 문제를 해결
* 또 다른 유용한 패턴은 확장형 메뉴나 능력 카드로 제공 기능을 구조적으로 열거하는 방식

* 개방형 텍스트 입력에 주로 의존하는 시스템에서는 에이전트가 대화 속에서 스스로 능력을 분명히 알릴 수 있어야 함
* 능력을 드러내는 것도 중요하지만 한 번에 너무 많은 옵션을 제시해 사용자를 압도하지 않는 것도 중요함

* 능력 커뮤니케이션은 에이전트가 무엇을 할 수 있는지 나열하는 데 그치지 않음
* 사용자가 그 능력을 자신 있게 효율적으로 활용할 수 있도록 경험을 설계하는 일

### 신뢰도와 불확실성 커뮤니케이션

* 불확실성을 효과적으로 전달하는 일은 사용자 신뢰를 쌓고 사용자가 정보에 근거한 결정을 내리도록 돕는 데 필수적

* 신뢰도의 표현 방식
    * 명시적 진술
        * '정답일 확률이 90%입니다.'
    * 시각적 신호
        * 아이콘, 색상 기반 경고 그래픽 인터페이스의 신뢰도 미터
    * 행동 조정
        * 신뢰도가 낮을 때 단정적 권고 대신 제안을 제공

* 에이전트는 불확실성이 높은 상황에서 과도한 자신감을 보여서는 안됨

### 사용자 지침과 입력 요청

* 아무리 고도화된 에이전트라도 모호하거나 애매하고 상충되는 사용자 입력을 완벽히 해석할 수 없음
* 위험한 추정을 하기보다 명확한 질문을 하거나 사용자 지침을 구할 시점을 알아야 함

* 효과적인 에이전트는 모호성에 직면했을 때 초점을 맞춘 유용한 질문을 하도록 설계됨
* 질문 방식도 중요함
* 왜 명확하가 필요한지 투명하게 설명해야 함
* 한 번에 너무 많은 질문을 피해야 함

### 우아한 실패

* 에이전트 시스템이 항상 제대로 작동한다는 법은 없음
* 실패를 다루는 방식은 성공만큼이나 중요함

* 우아한 실패의 핵심은 문제를 투명하게 인정하고 유용한 설명을 제공하며 실행 가능한 다음 단계를 제시하는 것
* 또한 에이전트는 흔한 실패 지점을 예측해 사전에 폴백 메커니즘을 갖추어야 함
* 다단계 작업에서는 실패 시 진행 상태 보존이 동일하게 중요함
* 사과와 공감의 언어도 핵심이며 명확한 해결 경로를 제공해야 함
* 가능하면 실패로부터 학습을 해야함

## 상호작용 설계에서의 신뢰

* 신뢰가 없으면 아무리 고도화된 에이전트도 사용자를 얻기 어려움
* 투명성과 예측 가능성은 신뢰를 구축, 유지하는 가장 강력한 도구

* 투명성은 에이전트의 능력과 제약을 명확히 전달하는 데서 시작
    * 사용자가 에이전트가 작업을 처리할 수 있는지 의도된 범위 안에서 작동하는지 추측하게 해서는 안됨
* 예측 가능성은 다양한 상황에서의 일관된 행동을 통해 투명성을 보완
    * 사용자는 과거 상호작용을 바탕으로 에이전트의 반응을 예측할 수 있어야 함

* 에이전트는 엣지 케이스도 신중히 처리해야 함
    * 예측 가능한 방식으로 대응해야 함

* 신뢰성의 또 다른 핵심은 시스템 복원력
    * 에이전트는 오류에서 회복하고 중단 상황에서도 상태를 유지하며 연쇄적으로 중단되지 않도록 설계해야 함
* 신뢰성은 기대를 일관되게 설정하고 충족하는 일
