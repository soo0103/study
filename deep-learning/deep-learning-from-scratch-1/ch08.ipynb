## 딥러닝

### 더 깊게

#### 더 깊은 신경망으로
* 층이 깊은 신경망을 구현
* 합성곱 계층은 모두 3 x 3 크기의 작은 필터로, 층이 깊어지면서 채널 수가 더 늘어나는 것이 특징
* 풀링 계층을 추가해 중간 데이터의 공간 크기를 저차 줄이고 마지막 단의 완전연결 계층에서는 드롭아웃 계층을 사용
* 가중치 초깃값으로 He 초깃값, 가중치 매개변수 갱신에는 Adam을 이용

#### 정확도를 더 높이려면
* MNIST 데이터셋에 대해서는 층을 아주 깊게 하지 않고도 최고 수준의 결과가 나옴
* 앙상블 학습, 학습률 감소, 데이터 확장 등은 정확도 향상에 도움됨
    * 데이터 확장
        * 입력 이미지를 알고리즘을 동원해 인위적으로 확장
        * 이미지를 회전하거나 세로로 이동하는 등
        * 일부를 잘라내는 crop, 좌우를 뒤집는 flip 등
        * 일반적인 이미지에는 밝기, 확대, 축소 등

#### 깊게 하는 이유
* ILSVRC로 대표되는 대규모 이미지 인식 대회의 결과에서 파악 가능
* 대부분 딥러닝 기반이 상위를 차지, 신경망을 더 깊게 만드는 방향

* 장점
    * 신경망의 매개변수 수가 줄어듦
    * 학습의 효율성이 높아짐 (학습 데이터의 양을 줄여 학습을 고속으로 수행)
    * 정보를 계층적으로 전달할 수 있음

### 딥러닝의 초기 역사

### 이미지넷
* ImageNet은 100만장이 넘는 이미지를 담고 있는 데이터셋
* 다양한 종류의 이미지를 포함하며 각 이미지에는 레이블(클래스 이름)이 붙어 있음

#### VGG
* 합성곱 계층과 풀링 계층으로 구성되는 기본적인 CNN
* 합성곱 계층, 완전연결 계층을 모두 16층(혹은 19층)으로 심화한게 특징
* 3 x 3의 작은 필터를 사용한 합성곱 계층을 연속으로 거침
* 합성곱 계층을 2~4회 연속으로 풀링 계층을 두어 크기를 절반으로 줄이는 처리를 반복

#### GoogLeNet
* 기본적으로는 CNN과 다르지 않음
* 단, 세로 방향 깊이와 가로 방향 깊이도 깊음
* 인셉션 구조: 가로 방향의 폭
    * 인셉션 구조는 다른 필터와 풀링을 여러 개 적용하여 그 결과를 결합
    * 이 구조를 하나의 구성요소로 사용하는 것이 특징
* 1 x 1 크기의 필터를 사용한 합성곱 계츠을 많은 곳에서 사용, 채널쪽으로 크기를 줄이는 것, 매개변수 제거와 고속 처리에 기여

#### ResNet
* 마이크로소프트 팀이 개발한 네트워크
* 스킵 연결을 도입 -> 층의 깊이에 비례해 성능을 향상시킬 수 있는 핵심  
    * 스킴 연결: 입력 데이터를 합성곱 계층을 건너뛰어 출력에 바로 더하는 구조
* 입력 x를 연속한 두 합성곱 계층을 건너뛰어 출력에 바로 연결 -> 출력이 F(x) + x가 됨
* 역전파 때 스킵 연결이 신호 감쇠를 막아줌
* ResNet은 합성곱 계층을 2개 층마다 건너뛰면서 층을 깊게 함

##### 전이 학습
* 학습된 가중치를 다른 신경망에 복사한 다음, 그 상태로 재학습을 수행
* 보유한 데이터셋이 적을 때 유용함

### 딥러닝 고속화
* 빅데이터와 네트워크의 발전으로 딥러닝에서는 대량의 연산을 수행해야 함  
    => 대부분 GPU로 대량의 연산을 고속으로 처리, GPU 또한 여러 개로 분산 수행

#### 풀어야 할 숙제
* AlexNet에서는 오랜 시간을 합성곱 계층에서 소요  
    => 합성곱 계층에서의 연산을 어떻게 고속으로 효율적으로 하느냐가 딥러닝의 과제

#### GPU를 활용한 고속화
* GPU는 병렬 수치 연산을 고속으로 처리할 수 있음  
    => 다양한 용도로 활용하자는 것이 GPU 컴퓨팅의 목적
* 딥러닝에서는 대량의 단일 곱셈 누산을 수행해야 함 -> GPU의 특기
* 대부분의 딥러닝 프레임워크는 엔비디아 GPU에서만 혜택
    => CUDA라는 엔비디아의 GPU 컴퓨팅용 통합 개발 환경을 사용
* cuDNN은 CUDA 위에서 동작하는 라이브러리

#### 분산 학습
* GPU로 딥러닝 연산을 가속 -> 그러나 오래 걸림  
    => 수평 확장(분산 학습)
* CNTK: 분산 학습에 초점을 둔 딥러닝 프레임워크
* GPU 수가 늘어날수록 학습도 빨라짐
* 계산을 어떻게 분산시키느냐는 어려운 문제

#### 연산 정밀도와 비트 줄이기
* 메모리 용량과 버스 대역폭 등이 딥러닝 고속화에 병목이 될 수 있음
* 메모리 용량 면에서는 대량의 가중치 매개변수와 중간 데이터를메모리에 저장해야 한다는 것을 생각해야 함
* 버스 대역폭 면에서는 버스를 흐르는 데이터가 한계를 넘어서면 병목이 됨  
    => 네트워크로 주고받는 데이터의 비트 수는 최소로 만드는 것이 좋음

* 딥러닝은 높은 수치 정밀도를 요구하지 않음
* 딥러닝은 16비트 반정밀도만 사용해도 학습에 문제 없음

### 딥러닝의 활용

#### 사물 검출
* 이미지 속에 담긴 사물의 위치와 종류를 알아내는 기술
* 사물 인식보다 어려우며 한 이미지에 여러 사물이 존재할 수도 있음
* CNN 기반으로 사물 검출하는 방식에 R-CNN이 유명함

* R-CNN의 흐름
1. 입력 이미지
2. 후보 영역 추출
3. CNN 특징 계산
4. 영역 분류
* Faster R-CNN 기법도 있음

#### 분할
* 이미지를 픽셀 수준에서 분류하는 문제
* 픽셀 단위로 객체마다 채색된 지도 데이터를 사용해 학습
* 추론할 때 입력 이미지의 모든 픽셀을 분류
* FCN 고안 -> 단 한 번의 forward 처리로 모든 픽셀의 클래스를 분류해주는 기법

* FCN은 완전연결 계층을 같은 기능을 하는 합성곱 계층으로 바꿈
* 공간 볼륨을 유지한 채 마지막 출력까지 처리 가능
* 마지막에 공간 크기를 확대하는 처리를 도입했음  
    -> 이중 선형 보간에 의한 선형 확대 (역합성곱 연산으로 구현)

#### 사진 캡션 생성
* CV + NLP
* 사진을 주면 그 사진을 설명하는 글(사진 캡션)을 자동으로 생성하는 연구
* NIC 모델이 대표적

* NIC는 심층 CNN과 자연어를 다루는 RNN으로 구성
    * RNN은 순환적 관계를 갖는 신경망으로 자연어나 시계열 데이터 등의 연속된 데이터를 다룰 때 많이 활용
* NIC는 CNN으로 사진에서 특징을 추출하고 특징을 RNN으로 넘김
* RNN은 CNN이 추출한 특징을 초깃값으로 텍스트를 순환적으로 생성  
=> 사진이나 자연어와 같은 여러 종류의 정보를 조합하고 처리하는 멀티모달 처리

### 딥러닝의 미래

#### 이미지 스타일(화풍) 변환
* 딥러닝으로 그림을 그리는 연구가 있음
* 두 이미지를 입력해 새로운 그림을 생성
* 하나는 콘텐츠 이미지, 다른 하나는 스타일 이미지
* 네트워크의 중간 데이터가 콘텐츠 이미지의 중간 데이터와 비슷해지도록 학습

#### 이미지 생성
* 아무런 입력 이미지 없이 새로운 이미지를 그려내는 연구
* 먼저 대량의 이미지를 사용해 학습, 학습 후에는 아무런 이미지 없이도 새로운 그림을 그림
* DCGAN 기법이 있음

* DCGAN은 생성자와 식별자로 불리는 2개의 신경망을 이용
* 생성자가 진짜와 똑같은 이미지를 생성, 식별자는 그것이 진짜인지 판정
* 생성자는 더 정교한 가짜 이미지 생성 기술을 학습하고 식별자는 더 정확하게 파악하도록 성장  
    => GAN 기술

#### 자율주행
* 주위 환경을 인식하는 기술에서 딥러닝이 큰 역할을 해줄 것
* SegNet이라는 CNN 기반 신경망은 주변 환경을 정확하게 인식

#### Deep Q-Network(강화학습)
* 강화학습에서는 에이전트가 환경에 맞게 행동을 선택하고 그 행동에 의해 환경이 변하는 것이 기본적인 틀
* 보상은 정해진 것이 아니라 예상 보상
* DQN라는 방법이 있음

* Q학습이라는 강화학습 알고리즘을 기초로 함
* 최적 행동 가치 함수로 최적인 해동을 정함
* 이 함수를 딥러닝으로 비슷하게 흉내내어 사용하는 것이 DQN
