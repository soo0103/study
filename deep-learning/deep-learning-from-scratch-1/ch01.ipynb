## Class

* class 키워드를 사용해 클래스를 정의
* `__init__`: 생성자
    * 클래스의 인스턴스가 만들어질 때 한 번만 불림
* 파이썬은 메서드의 첫 번째 인수로 자신(자신의 인스턴스)을 나타내는 self를 명시적으로 씀

```python
class Man:
    def __init__(self, name):
        self.name = name
        print("Initialized!")

    def hello(self):
        print("Hello " + self.name + "!")

    def goodbye(self):
        print("Good-bye " + self.name + "!")

m = Man("David")
m.hello()
m.goodbye()
```

## numpy

```python
import numpy as np

# 넘파이 배열
x = np.array([1.0, 2.0, 3.0])
print(x)
print(type(x))
```

```python
# 산술 연산
x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
```

* 산술 연산에서 x, y의 원소 수가 같음
* 원소 수가 동일하다면 산술 연산은 각 원소에 대해 행해짐
* `브로드캐스트`: 넘파이 배열과 수치 하나의 조합으로 된 산술 연산 시 스칼라값과 배열의 원소별로 한 번씩 수행

```python
# 브로드캐스트
x = np.array([1.0, 2.0, 3.0])
print(x / 2.0)
```

```python
# 다차원 배열
A = np.array([[1, 2], [3, 4]])

print(A)
print(A.shape)
print(A.dtype)
```

* shape: 행렬의 형상
* dtype: 행렬에 담긴 원소의 자료형

```python
B = np.array([[3, 0], [0, 6]])

print(A + B)
print(A * B)
```

* 형상이 같은 행렬이면 행렬의 산술 연산도 대응하는 원소별로 계산
* 마찬가지로 브로드캐스트도 가능

```python
print(A)
print(A * 10)
```

```python
# 브로드캐스트
A = np.array([[1, 2], [3, 4]])
B = np.array([10, 20])

print(A * B)
```

* 1차원 배열인 B가 2차원 배열 A와 똑같은 혀사으로 변형된 후 원소별 연산이 이루어짐

```python
# 원소 접근
X = np.array([[51, 55], [14, 19], [0, 4]])

print(X)
print(X[0])
print(X[0][1])
```

```python
for row in X:
    print(row)
```

```python
# 평탄화
print(X)
print(X.flatten()) # X를 1차원 배열로 변환
X = X.flatten()
print(X[np.array([0, 2, 4])]) # 인덱스가 0, 2, 4인 원소 얻기
```

```python
# 특정 조건
print(X > 15)
print(X[X > 15])
```

## matplotlib

```python
# 그래프 그리기
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 6, 0.1) # 0에서 6까지 0.1 간격으로 생성
y = np.sin(x)

plt.plot(y)
plt.show()
```

```python
# pyplot
x = np.arange(0, 6, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin")
plt.plot(x, y2, linestyle="--", label="cos")
plt.xlabel("x")
plt.ylabel("y")
plt.title("sin & cos")
plt.legend()
plt.show()
```
